MISSING VALUES

```python
# price per unit
df.isna().sum()
```

```python
df[df['Price Per Unit'].isna()]
```

```python
# to build a fixed price per item from rows were both item and price per unit are present
price_per_unit_lookup = (
    df.dropna(subset=['Item', 'Price Per Unit']).groupby('Item')['Price Per Unit'].first())

# fill missing price using the item lookup
df['Price Per Unit'] = df['Price Per Unit'].fillna(df['Item'].map(price_per_unit_lookup))
print('Remaining missing price:', df['Price Per Unit'].isna().sum())
```

```python
allowed_prices = price_per_unit_lookup.unique()
missing_price = df['Price Per Unit'].isna()
calculated_price = (df['Total Spent']/df['Quantity']).round(2)

# True where price is missing and calculated price matches an allowed price
matched = missing_price & calculated_price.isin(allowed_prices)
df.loc[matched, 'Price Per Unit'] = calculated_price[matched]
print('Filled prices:', matched.sum())
print('Still missing price:', df['Price Per Unit'].isna().sum())
```

```python
# checking whether category and price is unique to a single item
pair_counts = df.dropna(subset=['Category', 'Price Per Unit', 'Item']).groupby(['Category', 'Price Per Unit'])['Item'].nunique()
print('Is (category, Price per unit) unique per item?', (pair_counts == 1).all())

# fill missing items using the first known item for that (category, Price per unit)
before_missing = df['Item'].isna().sum()
group_item = df.groupby(['Category', 'Price Per Unit'])['Item'].transform('first')
df['Item'] = df['Item'].fillna(group_item)

# fill any still-missing item values with 'Unknown'
df['Item'] = df['Item'].fillna('Unknown')
recovered_count = before_missing - (df['Item'] == 'Unknown').sum()
print('Recovered_items:', recovered_count)
print('Total Unknown items:', (df['Item'] == 'Unknown').sum())
```

```python
# count rows where quantity and total is blank
(df['Quantity'].isna() & df['Total Spent'].isna()).sum()
```

```python
# count rows where only quantity or total is blank
(df['Quantity'].isna() ^ df['Total Spent'].isna()).sum()
```

```python
# drop rows where both quantity and total spent are missing
both_missing = df['Quantity'].isna() & df['Total Spent'].isna()
print('Dropping rows with both missing:', both_missing.sum())
df = df[~both_missing].reset_index(drop=True)
```

Missing values (after removing the 125 duplicates)

For the missing prices, Item was also missing on these rows, so i calculated Total Spent/Quantity and filled it only when the result matched a known price. All 607 passed.

For the missing items, each pair (Category & Price per unit) identifies exactly one item, so i recovered all of them from that pair(1,205 items). zero 'Unknown' rows were returned.

For the missing 598 Quantity and Total spent rows, they were always missing together, so neither could be derived from the other. I dropped the rows because they are negligible (about 5%) instead of inventing sales figures. The rows dropped from 12,450 to 11,852 rows.

For the missing discount Applied column( 3,953 rows), the values couldn't be inferred from any other column, so i left it as missing and kept the true/false data type.

OUTLIERS

```python
df.describe()
```

```python
columns = ['Price Per Unit', 'Quantity', 'Total Spent']
Q1 = df[columns].quantile(0.25)
Q3 = df[columns].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

print('Lower bounds:\n', lower_bound)
print('Upper bounds:\n', upper_bound)
```

```python
outliers = df[
    ((df[columns] < lower_bound) | (df[columns] > upper_bound)).any(axis=1)
    ]
outliers[columns]
```

```python
# count flagged rows and check if they're within range
print('Flagged IQR rows:', len(outliers))
out_of_range = df[
    ~df['Price Per Unit'].between(5, 41) | ~df['Quantity'].between(1,10)| (df['Total Spent'] > 410)]
print('Rows outside expected ranges (Price 5-41, Qty 1-10, Total <= 410:', len(out_of_range))
```

Price Per Unit and Quantity had no outliers. Total Spent had 60 rows flagged as outliers, all with Price 41.0 * Quantity 10 = 410.0, the largest possible purchase in the data.

A range check was carried out on the entire dataset and it confirmed no impossible or out-of-range values, meaning that the flagged rows are valid high-value orders, not errors, so i retained all 60.
