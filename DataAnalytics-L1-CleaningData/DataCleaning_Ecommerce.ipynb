Justification for the missing qauntities and Prices: 
Negative and missing values in the Quantity and Price columns were treated as mising values (NaN). I avoided imputing them with the mean or median because the actual quantities sold could not be established from the available information, and prices vary across individual products. Assigning estimated values without sufficient supporting information could distort sales calculations. These records were excluded from calculations requiring valid quantity and price values.

Outlier detection Using the Interquartile Range (IQR) Method

```python
# Identifying the outliers
columns = ['Quantity', 'Price', 'Calculated_Total']

Q1 = df_copy[columns].quantile(0.25)
Q3 = df_copy[columns].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)
print('Lower bounds:\n', lower_bound)
print('Upper bounds:\n', upper_bound)
```

```python
outliers = df_copy[
    ((df_copy[columns] < lower_bound) | (df_copy[columns] > upper_bound)).any(axis = 1)
]
outliers[columns]
```

```python
# to display the specific outlier rows
display(df_copy.loc[[17, 49, 94]])
```

```python
# imputing the blender price with the appropraite value
blender = df_copy[df_copy['Product'] == 'Blender']

# mean and median including the 10,000 outlier
print('MeanOut:', blender['Price'].mean())
print('MedianOut:', blender['Price'].median())

# mean and median without the 10,000 outlier
normal_blender = blender[blender['Price'] < 10000]
print('MeanW_Out:', normal_blender['Price'].mean())
print('MedianW_Out:', normal_blender['Price'].median())
```

```python
# replace the 10000 with the median blender price without the outlier
blender_median = normal_blender['Price'].median()
df_copy.loc[(df_copy['Product'] == 'Blender') & (df_copy['Price'] == 10000), 'Price'] = blender_median

# recalculate calculated_Total
df_copy['Calculated_Total'] = df_copy['Quantity'] * df_copy['Price']
```

```python
# check the fix
print('Max blender price now:', df_copy.loc[df_copy['Product'] == 'Blender', 'Price'].max())
```

Decision taken for Outliers:
A blender price of 10,000 was far above the other prices( the rest are under 1,000) and was treated as a data entry error. i replace it with the median blender price, since the median is not affected by extreme values. Calculated_Total was then recalculated.
